```python
from graph import build_graph
from langchain_core.messages import HumanMessage

app = build_graph()

solicitud = (
    "Investigá los pedidos de los clientes 102 y 205. "
    "Calculá cuánto gastó cada uno y determiná quién realizó "
    "compras por un importe mayor y cuál es la diferencia."
)

estado_inicial = {
    "messages": [HumanMessage(content=solicitud)],
    "next_agent": "research",
    "contributions": [],
    "steps": 0,
    "task_completed": False,
    "validation_passed": False,
}

print("=== DEMOSTRACIÓN DEL ORQUESTADOR ===\n")

for evento in app.stream(
    estado_inicial,
    stream_mode="updates",
    config={"recursion_limit": 30}
):
    for nodo, resultado in evento.items():
        print(f"Nodo ejecutado: {nodo}")

        if nodo == "supervisor":
            print(f"Delegación: {resultado['next_agent']}")

        if nodo == "validation":
            print(f"Validación aprobada: {resultado['validation_passed']}")

        if nodo == "synthesis":
            print(f"\nRESPUESTA FINAL:\n{resultado['messages'][-1].content}")

        print()
```
